```python
import numpy as np
import matplotlib.pyplot as plt


t = np.linspace(0,2,1000)
f =2
sine_wave = np.sin(2*np.pi*f*t)
plt.figure(figsize =(12,4))
plt.plot(t,sine_wave)
plt.title('sinewave')
plt.xlabel('time')
plt.ylabel('amplitude')
plt.grid()
plt.show()
```

```python
t = np.linspace(0, 2, 1000)
f = 2
cos_wave = np.cos(2*np.pi*f*t)

plt.plot(t, cos_wave)
plt.title('Cosine Wave')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```

```python
t = np.linspace(-1, 1, 1000)
step = np.heaviside(t, 1)

plt.plot(t, step)
plt.title('Unit Step Signal')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```

```python
t = np.linspace(0, 2, 1000)
exp_inc = np.exp(t)

plt.plot(t, exp_inc)
plt.title('Increasing Exponential')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
```

```python
t = np.linspace(0, 2, 1000)
exp_dec = np.exp(-t)

plt.plot(t, exp_dec)
plt.title('Decreasing Exponential')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```

```python
t = np.linspace(-1, 1, 1000)
ramp = np.maximum(0, t)

plt.plot(t, ramp)
plt.title('Ramp Signal')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```

```python
t = np.linspace(-1, 1, 1000)
impulse = np.zeros_like(t)
impulse[500] = 1  # spike at center

plt.stem(t, impulse)
plt.title('Unit Impulse (Approx)')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```

```python
n = np.arange(-10, 11)
impulse = np.zeros(len(n))
impulse[10] = 1  # n=0 position

plt.stem(n, impulse)
plt.title('Digital Unit Impulse')
plt.xlabel('n')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```

```python
t = np.linspace(-1, 1, 1000)
signum = np.sign(t)

plt.plot(t, signum)
plt.title('Continuous Signum Signal')
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```

```python
n = np.arange(-10, 11)
signum_d = np.sign(n)

plt.stem(n, signum_d)
plt.title('Discrete Signum Signal')
plt.xlabel('n')
plt.ylabel('Amplitude')
plt.grid()
plt.show()
```
